Importando bibliotecas 

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px 
```

Carregando DF

```python
df = pd.read_csv(r'C:\Users\Carlos\Desktop\Dashboard python\Dados\dataset_vendas.csv')
```

Primeiros e ultimos registros do DF

```python
df
```

Tratando os dados

```python
meses_map = {
    'Jan': int(1),
    'Fev': int(2),
    'Mar': int(3),
    'Abr': int(4),
    'Mai': int(5),
    'Jun': int(6),
    'Jul': int(7),
    'Ago': int(8),
    'Set': int(9),
    'Out': int(10),
    'Nov': int(11),
    'Dez': int(12)
}

df['Mês'] = df['Mês'].map(meses_map)
```

```python
df ['Valor Pago'] = df ['Valor Pago'].str.lstrip('R$ ').astype(int)
```

```python
status_map = {
    'Pago': int(1),
    'Não pago': int(0),
}

df['Status de Pagamento'] = df['Status de Pagamento'].map(status_map)
```

```python
df
```

Entendendo os tipos de dados

```python
df.info()
```

Vendas por equipe

```python
df1 = df.groupby('Equipe') ['Valor Pago'].sum().reset_index()
```

```python
df1
```

```python
fig1 = go.Figure(go.Bar(
    x=df1['Valor Pago'],
    y=df1['Equipe'],
    orientation='h',
    textposition = 'auto',
    text=df1['Valor Pago'],
    insidetextfont=dict(family='Times', size=12)
))
```

```python
fig1.show()
```

Chamadas Médias por dia do mês

```python
df2 = df.groupby('Dia') ['Chamadas Realizadas'].sum().reset_index()
```

```python
fig2 = go.Figure(go.Scatter(
    x=df2['Dia'],
    y=df2['Chamadas Realizadas'], 
    mode="lines",
    fill='tonexty',
))
```

```python
fig2.show()
```

```python
fig2.add_annotation(text='Chamadas Médias por dia do Mês',
                    xref="paper",
                    yref="paper",
                    font=dict(
                        size=20,
                        color='gray',
                    ),
                    align="center",
                    bgcolor='rgba(0,0,0,0.8)',
                    x=0.05,
                    y=0.85,
                    showarrow=False,
                    )

fig2.add_annotation(text=f'Média: {round(df2['Chamadas Realizadas'].mean(), 2)}',
                    xref="paper",
                    yref="paper",
                    font=dict(
                        size=30,
                        color='gray',
                    ),
                    align="center",
                    bgcolor='rgba(0,0,0,0.8)',
                    x=0.05,
                    y=0.55,
                    showarrow=False,
                    )
```
