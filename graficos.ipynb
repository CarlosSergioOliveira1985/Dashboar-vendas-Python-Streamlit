Tratando os dados

```python
meses_map = {
    'Jan': int(1),
    'Fev': int(2),
    'Mar': int(3),
    'Abr': int(4),
    'Mai': int(5),
    'Jun': int(6),
    'Jul': int(7),
    'Ago': int(8),
    'Set': int(9),
    'Out': int(10),
    'Nov': int(11),
    'Dez': int(12)
}

df['Mês'] = df['Mês'].map(meses_map)
```

```python
df ['Valor Pago'] = df ['Valor Pago'].str.lstrip('R$ ').astype(int)
```

```python
status_map = {
    'Pago': int(1),
    'Não pago': int(0),
}

df['Status de Pagamento'] = df['Status de Pagamento'].map(status_map)
```
